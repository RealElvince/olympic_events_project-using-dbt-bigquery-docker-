calculate mean of age, height and weight

```python
age_mean = int(df["Age"].mean())
height_mean = round(df["Height"].mean(),1)
weight_mean = round(df["Weight"].mean(),1)
```

```python
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Height'] = pd.to_numeric(df['Height'], errors='coerce')
df['Weight'] = pd.to_numeric(df['Weight'], errors='coerce')
```

```python
df['Age'] = df['Age'].fillna(age_mean).astype('Int64')      
df['Height'] = df['Height'].fillna(height_mean).astype(float)
df['Weight'] = df['Weight'].fillna(weight_mean).astype(float)
```
